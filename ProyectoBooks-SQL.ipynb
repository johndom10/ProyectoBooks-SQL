📌 *Este subconjunto puede ser utilizado para análisis filtrados donde se desee trabajar solo con publicaciones del siglo XXI.*

```python
query = '''
SELECT
    b.book_id,
    b.title,
    COUNT(DISTINCT r.review_id) AS review_count,
    ROUND(AVG(rt.rating), 2) AS avg_rating
FROM books b
LEFT JOIN reviews r ON b.book_id = r.book_id
LEFT JOIN ratings rt ON b.book_id = rt.book_id
GROUP BY b.book_id, b.title
ORDER BY review_count DESC;
'''

book_reviews_ratings = pd.read_sql(query, con=engine)

book_reviews_ratings.head(10)  
```

### ✅ Conclusión: Número de reseñas y calificación promedio por libro

Se realizó un análisis combinado de las tablas `reviews` y `ratings` para identificar qué libros tienen mayor participación por parte de los usuarios, tanto en forma de reseñas como de calificaciones.

```python
query = '''
SELECT
    p.publisher,
    COUNT(b.book_id) AS book_count
FROM books b
JOIN publishers p ON b.publisher_id = p.publisher_id
WHERE b.num_pages > 50
GROUP BY p.publisher
ORDER BY book_count DESC
LIMIT 1;
'''

top_publisher = pd.read_sql(query, con=engine)

top_publisher
```

### ✅ Conclusión: Editorial con más libros de más de 50 páginas

Se realizó un análisis para identificar la editorial que ha publicado la mayor cantidad de libros **con más de 50 páginas**, con el objetivo de filtrar folletos o publicaciones de poca extensión.

#### 📌 Resultado:
- 🏢 **Penguin Books** es la editorial líder, con **42 libros** publicados que superan las 50 páginas.

Este hallazgo es relevante porque:

- Permite enfocarse en editoriales que **apuestan por contenido más extenso y posiblemente de mayor profundidad**.
- Penguin Books puede representar un **socio clave** en futuras estrategias de contenido para la plataforma.
- Filtrar por número de páginas es una buena práctica para mejorar la **calidad de las recomendaciones y análisis**, excluyendo entradas poco significativas.


```python
query = '''
WITH rated_books AS (
    SELECT
        b.book_id,
        b.title,
        b.author_id,
        COUNT(r.rating_id) AS num_ratings,
        AVG(r.rating) AS avg_rating
    FROM books b
    JOIN ratings r ON b.book_id = r.book_id
    GROUP BY b.book_id
    HAVING COUNT(r.rating_id) >= 50
)

SELECT
    a.author,
    rb.title,
    ROUND(rb.avg_rating, 2) AS avg_rating,
    rb.num_ratings
FROM rated_books rb
JOIN authors a ON rb.author_id = a.author_id
ORDER BY rb.avg_rating DESC
LIMIT 1;
'''

top_rated_author = pd.read_sql(query, con=engine)

top_rated_author
```

### ✅ Conclusión: Autor con la calificación promedio más alta (mín. 50 calificaciones)

Se realizó un análisis para identificar al autor o autora cuyo libro tiene la **calificación promedio más alta**, considerando únicamente libros que hayan recibido **al menos 50 calificaciones**. Esta restricción asegura que el promedio sea estadísticamente confiable.

#### 📌 Resultado:

- 📖 **Libro:** *Harry Potter and the Prisoner of Azkaban*

```python
query = '''
WITH active_users AS (
    SELECT
        username
    FROM ratings
    GROUP BY username
    HAVING COUNT(rating_id) > 50
),
user_reviews AS (
    SELECT
        r.username,
        COUNT(rv.review_id) AS num_reviews
    FROM active_users r
    LEFT JOIN reviews rv ON r.username = rv.username
    GROUP BY r.username
)

SELECT
    ROUND(AVG(num_reviews), 2) AS avg_reviews_per_active_user
FROM user_reviews;
'''

avg_reviews = pd.read_sql(query, con=engine)

avg_reviews
```
